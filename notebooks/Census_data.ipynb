```python
VARS = [
    "B01003_001E",  # total population
    "B01002_001E",  # median age
    "B19013_001E",  # median household income
    "B17001_001E",  # poverty total
    "B17001_002E",  # poverty below
    "B15003_001E",  # education total (25+)
    "B15003_022E",  # bachelor's
    "B15003_023E",  # master's
    "B15003_024E",  # professional
    "B15003_025E",  # doctorate
]
```

```python
import requests
import pandas as pd

def fetch_census_year(year):
    url = f"https://api.census.gov/data/{year}/acs/acs5"

    params = {
        "get": ",".join(VARS),
        "for": "county:*",
        "in": "state:*",
        "key": API_KEY
    }

    r = requests.get(url, params=params)
    r.raise_for_status()

    data = r.json()

    # Must be a list of lists
    if not isinstance(data, list):
        raise ValueError(f"Unexpected Census response for {year}")

    return pd.DataFrame(data[1:], columns=data[0])
```

```python
dfs = []

for year in range(2012, 2023):
    print(f"Fetching Census {year}")
    df = fetch_census_year(year)
    df["year"] = year
    dfs.append(df)

census = pd.concat(dfs, ignore_index=True)
print("Final Census shape:", census.shape)
```

```python

# Load renamed Census data
census = pd.read_csv(
    r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\census_county_renamed_2012_2022.csv"
)

# Ensure numeric types
num_cols = [
    "population", "median_age", "median_income",
    "poverty_total", "poverty_below",
    "education_total_25plus",
    "bachelors_count", "masters_count",
    "professional_degree_count", "doctorate_count"
]
for c in num_cols:
    census[c] = pd.to_numeric(census[c], errors="coerce")

# Create rates (normalized features)
census["poverty_rate"] = census["poverty_below"] / census["poverty_total"]

census["higher_education_rate"] = (
    census["bachelors_count"]
    + census["masters_count"]
    + census["professional_degree_count"]
    + census["doctorate_count"]
) / census["education_total_25plus"]

# Keep only ML-relevant columns
census_features = census[
    [
        "year", "state", "county",
        "population", "median_age", "median_income",
        "poverty_rate", "higher_education_rate"
    ]
].copy()

# Create FIPS for merging (string-safe)
census_features["fips"] = (
    census["state"].astype(str).str.zfill(2)
    + census["county"].astype(str).str.zfill(3)
)

# Save compact Census feature file
census_features.to_csv(
    r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\census_ml.csv",
    index=False
)
```
