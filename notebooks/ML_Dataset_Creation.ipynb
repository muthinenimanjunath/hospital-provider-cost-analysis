##### Column Dropping Framework  

Columns should only be dropped if they fall into one of these categories:  

Identifiers (non-generalizable)  

Target Leakage Variables  

Redundant Derived Variables  

High Multicollinearity Components  

Non-Analytical Administrative Fields  

Structural Missingness / Data Quality Issues  

```python
base_dataset.columns.tolist()
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer

# Copy dataset
ml_dataset = base_dataset.copy()

# STEP 1: Drop identifier columns
ml_dataset.drop(columns=[
    'hospital_name',      # Hospital name is identifier
    'city',               # City is identifier
    'county',             # County is identifier
    'zip_code',           # Zip code is identifier
    'state_full_name'     # Duplicate of state_code
], inplace=True)


# STEP 2: Census variables start from 2012, so filter out earlier years
ml_dataset = ml_dataset[ml_dataset['year'] >= 2012]   # Census variables start from 2012

# STEP 3: Separate numeric & categorical columns

numeric_cols = ml_dataset.select_dtypes(include=['float64','int64']).columns
cat_cols = ml_dataset.select_dtypes(include=['object']).columns

# STEP 4: Fill numeric missing values 
# We will use median imputation for numeric columns, which is robust to outliers.
num_imputer = SimpleImputer(strategy='median')  
ml_dataset[numeric_cols] = num_imputer.fit_transform(ml_dataset[numeric_cols])

# STEP 5: Fill categorical missing values

cat_imputer = SimpleImputer(strategy='most_frequent')
ml_dataset[cat_cols] = cat_imputer.fit_transform(ml_dataset[cat_cols])

# Convert 'year' to integer type
ml_dataset['year'] = ml_dataset['year'].astype(int)

# STEP 6: Save ML dataset
ml_dataset.to_csv("ml_dataset.csv", index=False)

print("ML dataset created successfully")
print("Shape:", ml_dataset.shape)
```
