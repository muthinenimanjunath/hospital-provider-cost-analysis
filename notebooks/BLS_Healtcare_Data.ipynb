```python

# Standardize column names
bls.columns = bls.columns.str.strip()

# Create unified state_code

bls["state_code"] = bls["ST"]

bls.loc[bls["state_code"].isna(), "state_code"] = bls["PRIM_STATE"]

# Create unified occupation fields

bls["occ_code_std"] = bls["OCC_CODE"]
bls.loc[bls["occ_code_std"].isna(), "occ_code_std"] = bls["occ_code"]

bls["occ_title_std"] = bls["OCC_TITLE"]
bls.loc[bls["occ_title_std"].isna(), "occ_title_std"] = bls["occ_title"]

# Create unified employment & wage
bls["tot_emp_std"] = bls["TOT_EMP"]
bls.loc[bls["tot_emp_std"].isna(), "tot_emp_std"] = bls["tot_emp"]

bls["a_mean_std"] = bls["A_MEAN"]
bls.loc[bls["a_mean_std"].isna(), "a_mean_std"] = bls["a_mean"]

# Convert numeric fields
bls["tot_emp_std"] = pd.to_numeric(bls["tot_emp_std"], errors="coerce")
bls["a_mean_std"] = pd.to_numeric(bls["a_mean_std"], errors="coerce")

# Select REQUIRED columns only
bls_working = bls[
    [
        "YEAR",
        "state_code",
        "occ_code_std",
        "occ_title_std",
        "tot_emp_std",
        "a_mean_std"
    ]
].copy()

# Rename cleanly
bls_working = bls_working.rename(columns={
    "YEAR": "year",
    "occ_code_std": "occ_code",
    "occ_title_std": "occ_title",
    "tot_emp_std": "tot_emp",
    "a_mean_std": "a_mean"
})

# Drop rows missing essentials
bls_working = bls_working.dropna(
    subset=["year", "state_code", "occ_code", "occ_title"]
)

# Save standardized BLS table
bls_working.to_csv(
    r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\bls_working_standardized.csv",
    index=False
)

print("Standardized BLS working table created")
print("Shape:", bls_working.shape)
print(bls_working.head())
```

```python
import pandas as pd

# Load saved healthcare context table
bls_healthcare = pd.read_csv(
    r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\bls_healthcare_data.csv"
)

# Drop aggregate SOC codes to avoid double counting
bls_healthcare = bls_healthcare[
    ~bls_healthcare["occ_code"].isin(["29-0000", "31-0000"])
].copy()

# Define occupation grouping
def map_occupation_group(title):
    title = title.lower()

    if "physician" in title or "surgeon" in title:
        return "Physician"

    if "registered nurse" in title:
        return "RN"

    if any(x in title for x in [
        "nurse practitioner",
        "pharmacist",
        "therapist",
        "dietitian",
        "optometrist",
        "chiropractor"
    ]):
        return "Practitioner"

    if any(x in title for x in [
        "assistant",
        "aide",
        "technician"
    ]):
        return "Support"

    return "Other"

# Apply grouping
bls_healthcare["occupation_group"] = bls_healthcare["occ_title"].apply(map_occupation_group)

# Quick check
print(bls_healthcare["occupation_group"].value_counts())
bls_healthcare[["occ_code", "occ_title", "occupation_group"]].head(10)
```

```python
import pandas as pd

# Load grouped healthcare BLS data
bls = pd.read_csv(
    r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\bls_healthcare_grouped.csv"
)

# Keep only ML-relevant groups
bls = bls[bls["occupation_group"].isin(
    ["Physician", "RN", "Practitioner", "Support"]
)].copy()

# Employment-weighted wage
# (sum(a_mean × tot_emp)) / (sum(tot_emp))
bls["weighted_wage"] = bls["a_mean"] * bls["tot_emp"]

# Aggregate by year, state, and group
ml_long = (
    bls
    .groupby(["year", "state_code", "occupation_group"])
    .agg(
        total_wage=("weighted_wage", "sum"),
        total_emp=("tot_emp", "sum")
    )
    .reset_index()
)

# Compute weighted average wage
ml_long["avg_wage"] = ml_long["total_wage"] / ml_long["total_emp"]

# Keep clean ML columns
ml_long = ml_long[
    ["year", "state_code", "occupation_group", "avg_wage"]
]

# Save long-format ML table
ml_long.to_csv(
    r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\bls_ml_wage_long.csv",
    index=False
)

print("ML long-format wage table created")
print(ml_long.head())
```
