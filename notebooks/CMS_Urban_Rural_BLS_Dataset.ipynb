```python
import pandas as pd
import numpy as np
```

```python
hospital = pd.read_csv(r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\ML Modelling\hospital_kpi_ready_with_rucc.csv")
hospital.head()
```

```python
bls_healthcare = pd.read_csv(r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\bls_healthcare_grouped.csv")
bls_healthcare.head()
```

```python
# Merge BLS wages into CMS hospital data
hospital_with_wages = hospital.merge(
    bls_healthcare,
    on=["year", "state_code"],
    how="left"
)
print("Merged dataset shape:", hospital_with_wages.shape)
hospital_with_wages.head()
```

```python
# How many occupation groups per hospital-year?
hospital_with_wages["occupation_group"].value_counts()

# Check duplication logic
hospital_with_wages.groupby(
    ["year", "state_code", "hospital_name"]
).size().value_counts().head()
```
