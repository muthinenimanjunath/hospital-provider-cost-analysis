```python
import pandas as pd
import glob
import os

DATA_FOLDER = r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\BLS"
OUTPUT_FILE = r"C:\Users\manju\OneDrive\Desktop\Documents\Data Analytics\DA Semester 4\Capstone 2\Supplementary Data\bls_all_years_full_columns.csv"

def read_excel_safe(path):
    return pd.read_excel(
        path,
        engine="xlrd" if path.endswith(".xls") else "openpyxl"
    )

def process_file(path):
    # Extract year from filename (state_M2011_dl.xls)
    year = int(os.path.basename(path)[7:11])

    df = read_excel_safe(path)

    # Keep original column names, just strip spaces
    df.columns = df.columns.str.strip()

    # Add year column
    df["YEAR"] = year

    return df

# Find all BLS files
files = (
    glob.glob(os.path.join(DATA_FOLDER, "state_M*_dl.xls")) +
    glob.glob(os.path.join(DATA_FOLDER, "state_M*_dl.xlsx"))
)

print(f"Found {len(files)} BLS files")

# Concatenate all years (union of columns)
bls_all_years = pd.concat(
    [process_file(f) for f in sorted(files)],
    ignore_index=True,
    sort=False  # IMPORTANT: keeps union of columns
)

# Save combined dataset
bls_all_years.to_csv(OUTPUT_FILE, index=False)

print("All BLS files merged with ALL columns preserved")
print("Shape:", bls_all_years.shape)
print(bls_all_years.head())
```
